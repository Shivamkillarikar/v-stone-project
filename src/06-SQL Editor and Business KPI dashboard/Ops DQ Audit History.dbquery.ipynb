```sql
SELECT 
    check_name,
    status,
    failure_count,
    executed_at
FROM meridian_tnn.ops.dq_audit_log
WHERE DATE(executed_at) >= :p_start_date
ORDER BY executed_at DESC;




SELECT 
    '1. fact_store_returns' AS source,
    COUNT(*) AS row_count,
    ROUND(SUM(refunded_cash), 2) AS total_val
FROM meridian_tnn.gold.fact_store_returns

UNION ALL

SELECT 
    '2. returns_metric_view',
    COUNT(*),
    ROUND(SUM(total_return_value), 2)
FROM meridian_tnn.gold.returns_metric_view;
```
